# **1. CliffWalking**

```python
import gymnasium as gym

env = gym.make("CliffWalking-v1")
episodes = 10
terminated = False

for episode in range(episodes):
    state, info = env.reset()
    total_reward = 0

    while not terminated:
        action = env.action_space.sample()
        next_state, reward, terminated, truncated, info = env.step(action)
        total_reward += reward

    print(f"total reward {total_reward} for episode {episode}")
    terminated = False

env.close()
```

# **2. Taxi**

```python
import gymnasium as gym

env = gym.make("Taxi-v4")

state, info = env.reset()
current_state = state

episodes = 10
total_reward = 0

for episode in range(episodes):
    action = env.action_space.sample()
    next_state, reward, terminated, truncated, info = env.step(action)

    print(f"episode={episode+1} | next_state={next_state} | current_state={current_state} | reward={reward} | terminated={terminated} | truncated={truncated} | action={action}")

    total_reward += reward

    if terminated or truncated:
        break

    current_state = next_state

print(f"Total reward={total_reward}")
print(f"Average reward={total_reward/(episode+1)}")

env.close()
```
