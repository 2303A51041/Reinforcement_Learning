```python

gamma = 0.9

states = {
    "S0": (0, 0),
    "S1": (0, 1),
    "G": (0, 2),
    "X": (1, 0),
    "S3": (1, 1),
    "S4": (1, 2)
}

policy = {
    "S0": "Right",
    "S1": "Down",
    "S3": "Right",
    "S4": "Up"
}

actions = ["Up", "Down", "Left", "Right"]

def step(state, action):
    row, col = states[state]

    moves = {
        "Up": (-1, 0),
        "Down": (1, 0),
        "Left": (0, -1),
        "Right": (0, 1)
    }

    dr, dc = moves[action]
    nr, nc = row + dr, col + dc

    if nr < 0 or nr > 1 or nc < 0 or nc > 2:
        return state, -1

    next_state = next(
        s for s, pos in states.items()
        if pos == (nr, nc)
    )

    if next_state == "X":
        return next_state, -5

    if next_state == "G":
        return next_state, 10

    return next_state, -1

def state_value_function(state):
    total_return = 0
    discount = 1
    current_state = state

    while current_state not in ["G", "X"]:
        action = policy[current_state]
        next_state, reward = step(current_state, action)

        total_return += discount * reward
        discount *= gamma
        current_state = next_state

    return total_return

def action_value_function(state, action):
    next_state, reward = step(state, action)

    if next_state in ["G", "X"]:
        return reward

    return reward + gamma * state_value_function(next_state)

print("State Value V(S0):", round(state_value_function("S0"), 4))
print("Action Value Q(S0, Up):", round(action_value_function("S0", "Up"), 4))
```
