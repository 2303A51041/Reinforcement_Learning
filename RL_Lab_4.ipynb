```python
import math

# Machine Initialization
machines = ['A', 'B', 'C']

# Number of plays
N_play = 7

# Initialize N, R, Q and UCB
N = [0, 0, 0]
R = [0, 0, 0]
Q = [0.0, 0.0, 0.0]
UCB = [0.0, 0.0, 0.0]

# Static rewards for 7 plays
rewards = [3, 6, 4, 7, 5, 8, 6]

# Play for 7 times
for t in range(1, N_play + 1):

    # Calculate UCB
    for i in range(3):
        if N[i] == 0:
            UCB[i] = float('inf')
        else:
            UCB[i] = Q[i] + math.sqrt((2 * math.log(t)) / N[i])

    # Select machine with highest UCB
    selected = UCB.index(max(UCB))

    # Get static reward
    reward = rewards[t - 1]

    # Update count
    N[selected] += 1

    # Update total reward
    R[selected] += reward

    # Update Q-value
    Q[selected] = Q[selected] + \
                  (reward - Q[selected]) / N[selected]

    print("Play", t,
          "Machine:", machines[selected],
          "Reward:", reward)

# Display final results
print("\nFinal Results")
print("-------------------------")

for i in range(3):
    print("Machine:", machines[i])
    print("N =", N[i])
    print("R =", R[i])
    print("UCB =", round(UCB[i], 3))
    print("Q =", round(Q[i], 3))
    print()

print("Total Reward =", sum(R))
```
