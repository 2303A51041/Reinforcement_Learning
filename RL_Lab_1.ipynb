1ST QUESTION

```python
# Multi-Armed Bandit - Greedy Algorithm
# Using Arithmetic Mean to estimate Q(a)

# Take number of machines
n = int(input("Enter number of machines: "))

machines = {}

# Take rewards for each machine
for i in range(n):
    name = input(f"\nEnter name of Machine {i+1}: ")
    rewards = list(map(float, input(f"Enter rewards for Machine {name} separated by spaces: ").split()))
    machines[name] = rewards

# Calculate Q(a) using arithmetic mean
print("\n----- Estimated Q Values -----")

q_values = {}

for machine, rewards in machines.items():
    q = sum(rewards) / len(rewards)
    q_values[machine] = q

    print(f"\nMachine {machine}")
    print(f"Rewards: {rewards}")
    print(f"Q({machine}) = {q:.2f}")

# Greedy selection
selected_machine = max(q_values, key=q_values.get)

print("\n------------------------------")
print(f"Greedy Selected Machine: {selected_machine}")
print(f"Highest Q value: {q_values[selected_machine]:.2f}")
```

2ND QUESTION

```python
# Multi-Armed Bandit - Greedy Algorithm
# Using Incremental Update Method

# Number of machines
n = int(input("Enter number of machines: "))

machines = {}

# Initialize Q and N
for i in range(n):
    name = input(f"Enter name of Machine {i+1}: ")
    rewards = list(map(float, input(f"Enter rewards for Machine {name}: ").split()))

    machines[name] = {
        "rewards": rewards,
        "N": 0,
        "Q": 0.0
    }

print("\n===== Incremental Update Process =====")

# Process every reward
for machine, data in machines.items():

    print(f"\nMachine {machine}")

    for reward in data["rewards"]:

        # Increment number of plays
        data["N"] += 1

        # Previous Q value
        old_q = data["Q"]

        # Incremental update formula
        data["Q"] = old_q + (reward - old_q) / data["N"]

        print(
            f"Reward = {reward:.2f}, "
            f"N = {data['N']}, "
            f"Q = {data['Q']:.2f}"
        )

# Display final values
print("\n===== Final Q Values =====")

q_values = {}

for machine, data in machines.items():
    q_values[machine] = data["Q"]

    print(f"Machine {machine}")
    print(f"Number of plays N({machine}) = {data['N']}")
    print(f"Q({machine}) = {data['Q']:.2f}")
    print()

# Greedy selection
selected_machine = max(q_values, key=q_values.get)

print("==========================")
print(f"Greedy Selected Machine: {selected_machine}")
print(f"Highest Q Value: {q_values[selected_machine]:.2f}")
```
