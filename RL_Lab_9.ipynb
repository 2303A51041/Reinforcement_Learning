## Assignment 1: Algebraic Solution

```python
gamma = 0.9

# Bellman equations:
# V(S0) = -1 + 0.81 * V(S1)
# V(S1) = 89.9 + 0.09 * V(S0)

# Solve the simultaneous equations algebraically
V0 = (-1 + 0.81 * 89.9) / (1 - 0.81 * 0.09)
V1 = 89.9 + 0.09 * V0

Q_S0_Down = -1 + gamma * V1
Q_S0_Right = -1
Q_S1_Right = 100
Q_S1_Up = -1 + gamma * V0

print("Assignment 1: Algebraic Solution")
print(f"V(S0) = {V0:.2f}")
print(f"V(S1) = {V1:.2f}")
print(f"Q(S0, Down) = {Q_S0_Down:.2f}")
print(f"Q(S0, Right) = {Q_S0_Right:.2f}")
print(f"Q(S1, Right) = {Q_S1_Right:.2f}")
print(f"Q(S1, Up) = {Q_S1_Up:.2f}")

print("\nVerification")
print(f"V(S0) from policy = {0.9 * Q_S0_Down + 0.1 * Q_S0_Right:.2f}")
print(f"V(S1) from policy = {0.9 * Q_S1_Right + 0.1 * Q_S1_Up:.2f}")
```

## Assignment 2: Iterative Policy Evaluation

```python
gamma = 0.9
theta = 0.000001

V0 = 0.0
V1 = 0.0
iterations = 0

while True:
    new_V0 = 0.1 * (-1 + gamma * 0) + 0.9 * (-1 + gamma * V1)
    new_V1 = 0.9 * (100 + gamma * 0) + 0.1 * (-1 + gamma * V0)

    delta = max(abs(new_V0 - V0), abs(new_V1 - V1))
    V0 = new_V0
    V1 = new_V1
    iterations += 1

    if delta < theta:
        break

Q_S0_Down = 0.1 * (-1 + gamma * 0) + 0.9 * (-1 + gamma * V1)
Q_S0_Right = -1 + gamma * 0
Q_S1_Right = 0.9 * (100 + gamma * 0) + 0.1 * (-1 + gamma * V0)
Q_S1_Up = -1 + gamma * V0

print("Assignment 2: Iterative Policy Evaluation")
print("Number of iterations:", iterations)
print(f"V(S0) = {V0:.2f}")
print(f"V(S1) = {V1:.2f}")

print("\nAction Values")
print(f"Q(S0, Down) = {Q_S0_Down:.2f}")
print(f"Q(S0, Right) = {Q_S0_Right:.2f}")
print(f"Q(S1, Right) = {Q_S1_Right:.2f}")
print(f"Q(S1, Up) = {Q_S1_Up:.2f}")

print("\nPolicy Consistency Checks")
print(f"V(S0) = Q(S0, Down): {abs(V0 - Q_S0_Down) < theta}")
print(f"V(S1) = Q(S1, Right): {abs(V1 - Q_S1_Right) < theta}")
```
