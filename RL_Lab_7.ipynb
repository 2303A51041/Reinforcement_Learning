**1. CartPole**

```python
import gymnasium as gym

env = gym.make("CartPole-v1")

print("Environment created")

current_state, info = env.reset()

print("=" * 50)
print("Initial State Information")
print("=" * 50)
print("Cart position:", current_state[0])
print("Cart velocity:", current_state[1])
print("Pole angle:", current_state[2])
print("Pole velocity:", current_state[3])
print("=" * 50)

episodes = 10
total_reward = 0

for episode in range(episodes):
    current_state, info = env.reset()
    total_reward = 0

    while True:
        action = env.action_space.sample()

        if action == 0:
            print("Cart moves left")
        if action == 1:
            print("Cart moves right")

        next_state, reward, terminated, truncated, info = env.step(action)

        print("=" * 50)
        print("Next State Information")
        print("=" * 50)
        print("Next Cart position:", next_state[0])
        print("Next Cart velocity:", next_state[1])
        print("Next pole angle:", next_state[2])
        print("Next Pole velocity:", next_state[3])
        print("Reward:", reward)
        print("=" * 50)

        total_reward += reward

        print(
            f"episode={episode + 1} | reward={reward} | "
            f"current_state={current_state} | next_state={next_state}"
        )

        if terminated or truncated:
            break

        current_state = next_state

    print("Total reward=", total_reward)
    print("Average reward=", total_reward / episodes)

env.close()
```

**2. MountainCar**

```python
import gymnasium as gym

env = gym.make("MountainCar-v0")

current_state, info = env.reset()

print("=" * 50)
print("Current State Information: Reset Environment")
print("=" * 50)
print(f"Car Position={current_state[0]}")
print(f"Car Velocity={current_state[1]}")
print("=" * 50)

episodes = 10
total_reward = 0

for episode in range(episodes):
    current_state, info = env.reset()
    total_reward = 0

    while True:
        action = env.action_space.sample()

        if action == 0:
            print("Accelerate left direction")
        if action == 1:
            print("No Accelerate")
        if action == 2:
            print("Accelerate right direction")

        next_state, reward, terminated, truncated, info = env.step(action)

        print(f"episode={episode+1} | next_state={next_state} | current_state={current_state} | reward={reward} | terminated={terminated}")

        total_reward += reward

        if terminated or truncated:
            break

        current_state = next_state

    print("Total reward=", total_reward)
    print("Average reward=", total_reward / (episode + 1))

env.close()
```

**3. FrozenLake**

```python
import gymnasium as gym

env = gym.make("FrozenLake-v1")

current_state, info = env.reset()
print("current_state:", current_state)

total_reward = 0
episodes = 10

for episode in range(episodes):
    action = env.action_space.sample()
    next_state, reward, terminated, truncated, info = env.step(action)

    print(f"episode={episode+1} | next_state={next_state} | current_state={current_state} | reward={reward} | terminated={terminated} | truncated={truncated} | action={action}")

    total_reward = total_reward + reward
    current_state = next_state

    if terminated or truncated:
        break

    average_reward = total_reward / (episode + 1)
    print(f"total_reward={total_reward} | average_reward={average_reward}")

env.close()
```
