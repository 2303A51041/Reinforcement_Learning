```python
import random

# Part A: Machine initialization
machines = ["A", "B", "C"]

# N = number of plays
# R = total reward
# Q = estimated average reward
N = {machine: 0 for machine in machines}
R = {machine: 0.0 for machine in machines}
Q = {machine: 0.0 for machine in machines}

epsilon = 0.2
total_reward = 0.0

# Static reward patterns given in the practical
rewards = {
    "A": [3, 4, 2, 5],
    "B": [6, 7, 5, 8],
    "C": [4, 3, 5, 4]
}

# Track how many times each machine has been selected
reward_index = {
    "A": 0,
    "B": 0,
    "C": 0
}

print("Modified Epsilon-Greedy Multi-Armed Bandit")
print("Number of plays: 10")
print("Epsilon:", epsilon)

# Part C: Modified Epsilon-Greedy Decision
for play in range(1, 11):

    # Find highest Q-value
    highest_q = max(Q.values())

    # Find machines having highest Q-value
    best_machines = [
        m for m in machines
        if Q[m] == highest_q
    ]

    # Randomly select one if there is a tie
    best_machine = random.choice(best_machines)

    # Generate random number
    r = random.random()

    if r < epsilon:
        # Exploration
        other_machines = [
            m for m in machines
            if m != best_machine
        ]

        selected_machine = random.choice(other_machines)
        decision = "Exploration"

    else:
        # Exploitation
        selected_machine = best_machine
        decision = "Exploitation"

    # Part D: Static Reward Collection

    index = reward_index[selected_machine]

    # Repeat reward pattern if necessary
    reward = rewards[selected_machine][
        index % len(rewards[selected_machine])
    ]

    # Increase reward index
    reward_index[selected_machine] += 1

    # Part E: Q-Value Update
    N[selected_machine] += 1
    R[selected_machine] += reward

    Q[selected_machine] = Q[selected_machine] + (
        reward - Q[selected_machine]
    ) / N[selected_machine]

    total_reward += reward

    print("\nPlay", play)
    print("Decision:", decision)
    print("Selected Machine:", selected_machine)
    print("Reward:", reward)


# Part F: Display Final Results
print("\n--- Final N, R, Q Tables ---")
print("Machine\tN (plays)\tR (total reward)\tQ (average reward)")

for machine in machines:
    print(
        f"{machine}\t"
        f"{N[machine]}\t\t"
        f"{R[machine]:.2f}\t\t\t"
        f"{Q[machine]:.2f}"
    )

print("\nTotal reward received by the player:", total_reward)
```
